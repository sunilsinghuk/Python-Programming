### **Creating a Pie Chart**


```python
import matplotlib.pyplot as plt
import numpy as np

# Generate random data for the pie chart
np.random.seed(42)  # For reproducibility
values = np.random.randint(10, 50, 5)  # Generate 5 random values
labels = ['A', 'B', 'C', 'D', 'E']  # Categories

# Create a pie chart
plt.figure(figsize=(6, 6))  # Set figure size
plt.pie(values, labels=labels, autopct='%1.1f%%', startangle=90, colors=['blue', 'green', 'red', 'orange', 'purple'])

# Add a title
plt.title("Random Data Pie Chart")

# Show the plot
plt.show()
```

**Explanation**
  * **np.random.randint(10, 50, 5):** Generates 5 random values between 10 and 50.

  * **labels=['A', 'B', 'C', 'D', 'E']:** Defines category labels.

  * **autopct='%1.1f%%':** Displays percentage values inside the pie chart.

  * **startangle=140:** Rotates the pie chart for better visualization.

  * **colors=['blue', 'green', 'red', 'orange', 'purple']:** Custom colors for each slice.

## **Lecture 3: Working with Multiple Plots (1 Hour)**
### **3.1 Creating Subplots**

```python
fig, ax = plt.subplots(2, 2, figsize=(8, 6))

x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

ax[0, 0].plot(x, y, marker ="^", color='red')
ax[0, 0].set_title("Subplot 1")

ax[0, 1].bar(x, y, color='blue')
ax[0, 1].set_title("Subplot 2")

ax[1, 0].scatter(x, y, color='green')
ax[1, 0].set_title("Subplot 3")

ax[1, 1].pie(y, labels=x, autopct='%1.1f%%')
ax[1, 1].set_title("Subplot 4")

plt.tight_layout()
plt.show()
```

# **Lecture 5: Plotting Categorical Data**
## **Bar Chart**
Bar plots are useful for visualizing categorical data by displaying the quantity of each category.

## **Vertical Bar Plot Example**

```python
categories = ['A', 'B', 'C', 'D']
values = [5, 7, 3, 8]

plt.bar(categories, values, color='green')
plt.title("Bar Chart Example")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

```python
import matplotlib.pyplot as plt

categories = ['Apples', 'Bananas', 'Cherries']
values = [10, 24, 36]

plt.bar(categories, values, color='skyblue')
plt.title('Fruit Sales')
plt.xlabel('Fruit')
plt.ylabel('Quantity')
plt.show()
```

**Horizontal Bar Plot Example**


```python
plt.barh(categories, values, color='lightgreen')
plt.title('Fruit Sales (Horizontal)')
plt.xlabel('Quantity')
plt.ylabel('Fruit')
plt.show()
```

### **🔹 Adding Error Bars**
Error bars indicate the uncertainty of data.

```python
import numpy as np

values = [20, 35, 30]
errors = [2, 3, 4]

plt.bar(categories, values, yerr=errors, capsize=5, color='salmon')
plt.title('Sales with Error Bars')
plt.xlabel('Fruit')
plt.ylabel('Sales')
plt.show()
```

### **🔹 Stacked Bar Charts**
Used to show parts of a whole for each category.

```python
men = [20, 35, 30]
women = [25, 32, 34]

x = np.arange(len(categories))

plt.bar(x, men, label='Men', color='blue')
plt.bar(x, women, bottom=men, label='Women', color='orange')

plt.xticks(x, categories)
plt.xlabel('Fruit')
plt.ylabel('Quantity')
plt.title('Fruit Sales by Gender')
plt.legend()
plt.show()
```

### **🔹 Grouped Bar Charts**
Used to compare multiple groups.

```python
bar_width = 0.35
x = np.arange(len(categories))

plt.bar(x, men, width=bar_width, label='Men', color='blue')
plt.bar(x + bar_width, women, width=bar_width, label='Women', color='orange')

plt.xticks(x + bar_width / 2, categories)
plt.xlabel('Fruit')
plt.ylabel('Quantity')
plt.title('Grouped Bar Chart')
plt.legend()
plt.show()
```

### **8.2 Heatmap**

```python
import pandas as pd

# Sample Data
data = {
    'Math': [85, 90, 78, 92, 88],
    'Science': [89, 94, 76, 95, 90],
    'English': [78, 85, 80, 86, 83],
    'History': [82, 87, 79, 90, 85]
}

# Create DataFrame
df = pd.DataFrame(data)

# Create correlation matrix
corr_matrix = df.corr()

# Plot heatmap
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title("Subject Correlation Heatmap")
plt.show()
```

# 8.3 Pairplots (Exploring Relationships)
The pairplot() function in Seaborn creates pairwise scatter plots and histograms for numerical columns.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load a built-in dataset
df = sns.load_dataset('iris')

# Pairplot with species as hue
sns.pairplot(df, hue='species', diag_kind='kde')
plt.suptitle('Pairplot of Iris Dataset', y=1.02)
plt.show()
```

**📌 Explanation:**

  * hue: Differentiates data by color based on category.

  * diag_kind: Chooses between histogram ('hist') and KDE ('kde') for diagonal plots.

### **What is hue in Seaborn?**
The hue parameter adds color coding to your plots based on the values of a categorical variable. It helps in grouping data visually.

### **What is kde in Seaborn?**
kde stands for Kernel Density Estimate. It adds a smoothed curve over a histogram or along the diagonal of pairplots to show the distribution shape.

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

# Plot total bill vs tip, color-coded by gender
sns.scatterplot(x='total_bill', y='tip', data=tips, hue='sex')
plt.title("Scatter Plot with Hue")
plt.show()
```

**Explanation:**

* hue='sex': Colors the points based on male/female.

* Useful in visualizing how different groups behave.

```python
sns.histplot(tips['total_bill'], bins=20, kde=True)
plt.title("Histogram with KDE Curve")
plt.show()
```

Explanation:

* kde=True: Adds a smooth curve on top of the histogram to better understand the data distribution.

* The KDE is like a continuous version of a histogram.

# **8.4 Heatmaps (Correlation Matrix)**
A heatmap shows the correlation or magnitude between variables using colors.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load Titanic dataset
titanic = sns.load_dataset('titanic')

# Compute correlation on numerical columns
corr = titanic.corr(numeric_only=True)

# Create heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap (Titanic Dataset)')
plt.show()
```

**Explanation:**

* annot=True: Displays correlation values.

* cmap: Defines color map.

* linewidths: Adds lines between cells for readability.

# **Visualizing Categorical Data**
## **9.1 Bar Plot**


```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')
sns.barplot(x='day', y='total_bill', data=tips, estimator=sum, errorbar=None)
plt.title('Total Bill by Day')
plt.show()
```

## **9.2 Count Plot (Frequency)**
Count plots show the count of observations in each category.

```python
sns.countplot(x='day', data=tips, palette='Set2', hue='day', legend=True)
plt.title('Count of Customers by Day')
plt.show()
```

## **9.3 Box Plot (Statistical Summary)**
Box plot shows median, quartiles, and outliers.

```python
sns.boxplot(x='day', y='total_bill', data=tips)
plt.title('Box Plot of Total Bill by Day')
plt.show()
```

## **9.4 Violin Plot (Distribution + KDE)**
Violin Plot combines box plot and KDE to show distribution shape.

```python
sns.violinplot(x='day', y='total_bill', data=tips, hue='sex', split=True)
plt.title('Violin Plot of Total Bill by Day and Gender')
plt.show()
```

# **Plot Aesthetics and Combining Plots**
## **10.1 Customizing Themes**

```python
sns.set_style('darkgrid')  # Options: white, dark, whitegrid, darkgrid, ticks
sns.set_palette('pastel')  # Palette can be customized
```

**Example: Custom Theme on Plot**

```python
sns.set_style('whitegrid')
sns.boxplot(x='day', y='tip', data=tips, palette='coolwarm', hue='day', legend=True)
plt.title('Customized Box Plot')
plt.show()
```

# **Combining Seaborn and Matplotlib Plots**
Overlaying multiple plot types.

```python
plt.figure(figsize=(8, 6))
sns.boxplot(x='day', y='total_bill', data=tips, whis=[0, 100], width=.6, palette="vlag")
sns.stripplot(x='day', y='total_bill', data=tips, size=4, color=".3", linewidth=0)
plt.title('Boxplot with Stripplot Overlay')
plt.show()
```

# **Combining Multiple Plots in One Figure**


```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.boxplot(x='day', y='total_bill', data=tips, ax=axes[0])
axes[0].set_title('Box Plot')

sns.violinplot(x='day', y='total_bill', data=tips, ax=axes[1])
axes[1].set_title('Violin Plot')

plt.tight_layout()
plt.show()
```
