# **PANDAS LIBRARY**

##**1. Introduction to Pandas**

**What is Pandas?**
1. Definition: Pandas is an open-source Python library designed for data manipulation and analysis. It provides flexible and powerful data structures such as Series and DataFrames to work with structured data.
2. Developed By: Wes McKinney in 2008.
3. Name Origin: The name "Pandas" is derived from "Panel Data", a term for multidimensional structured data sets.

**Key Features of Pandas**
1. Easy Handling of Missing Data: Built-in methods for detecting and handling missing values.
2. Flexible Data Structures:
  *   Series (1D)
  *   DataFrame (2D)

3. Data Manipulation: Filtering, grouping, merging, reshaping, and aggregation.
4. I/O Tools: Read/write data from various file formats (CSV, Excel, JSON, SQL, etc.).
5. Integration: Works seamlessly with other libraries like NumPy, Matplotlib, and Scikit-learn.

**Common Use Case**

*   Data cleaning and preparation.
*   Exploratory data analysis (EDA).
*   Feature engineering for machine learning.

## **2. Series**

**What is a Pandas Series?**

A Series is a 1-dimensional labeled array capable of holding any data type (integers, floats, strings, etc.).

Each element in a Series is associated with an index, allowing easy access to elements.

###**Creating a Pandas Series**
**1. From a List:**

```python
import pandas as pd
data = [10, 20, 30, 40]
series = pd.Series(data)
print(series)
```

**2. From a Dictionary:**

```python
data = {'a': 1, 'b': 2, 'c': 3}
series = pd.Series(data)
print(series)
```

**3. From a NumPy Array:**


```python
import numpy as np
data = np.array([1, 2, 3, 4])
series = pd.Series(data)
print(series)
```

###**Accessing Data in a Series**
**1. Indexing:** Access elements by position or label.

```python
print(series[0])  # By position
#print(series['a'])  # By label
```

**2. Slicing:** Extract multiple elements.

```python
print(series[1:3])  # Slicing by position
```

##**3. DataFrames**
**What is a Pandas DataFrame?**

A DataFrame is a 2-dimensional labeled data structure. It’s like a table in a relational database or a spreadshee.

It consists of rows (indexed) and columns (labeled).

###**Creating a Pandas DataFrame**
**1. From a Dictionary:**



```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Los Angeles', 'Chicago']
}
df = pd.DataFrame(data)
print(df)
```

**2. From a List of Lists:**

```python
data = [
    ['Alice', 25, 'New York'],
    ['Bob', 30, 'Los Angeles'],
    ['Charlie', 35, 'Chicago']
]
df = pd.DataFrame(data, columns=['Name', 'Age', 'City'])
print(df)
```

**3. From a CSV File:**

**Use read_csv()** to load data from a CSV file.

```python
#df = pd.read_csv('data.csv') #Syntax for Data Read, we need to pass data file link or path with in round bracket()
print(df.head())  # Display the first 5 rows
```

###**Basic DataFrame Operations**
**1. Access Columns:**

```python
print(df['Name'])  # Access a single column
print(df[['Name', 'City']])  # Access multiple columns
```

:**2. Access Rows:**

*   By index position:
*   By label:



```python
print(df.iloc[0])  # First row
print(df.loc[0])  # First row
```
