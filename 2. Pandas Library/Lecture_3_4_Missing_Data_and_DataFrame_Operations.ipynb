# **1. Introduction to Missing Data**
- Understanding missing data (NaN)
- Reasons for missing data

###**What is Missing Data?**
Definition: Missing data refers to the absence of a value in a dataset. In Pandas, missing data is represented as NaN (Not a Number).

###**Reasons for Missing Data**
**1. Data collection errors:** Sensors or tools failed to record data.    
**2. Human errors:** Missing entries in forms or surveys.  
**3. Data extraction issues:** Missing values during scraping or migration.  
**4. Intentional omission**: Certain data points might not be applicable to all records.

```python
import pandas as pd
import numpy as np

# Example DataFrame with missing values
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, np.nan, 30, 29],
    'City': ['New York', 'Los Angeles', np.nan, 'Chicago']
}
df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
```

# **2. Handling Missing Data (30 mins)**
###**Identifying Missing Data**
- Use isnull() to detect missing values.
- Use notnull() to find non-missing values.
- Both methods return a Boolean DataFrame.


### **Identifying Missing Data**
- Using `isnull()` and `notnull()` to identify missing data

```python
# Identifying missing data
print(df.isnull())  # True for missing values
print(df.notnull())  # True for non-missing values

# Count missing values per column
print(df.isnull().sum())
```

### **Dropping Missing Values**
- Dropping rows with missing values
- Dropping columns with missing values

###**Dropping Missing Values**
- Use dropna() to remove rows or columns with missing values.
- Parameters:  
 - axis=0: Drop rows (default).
 - axis=1: Drop columns.
 - how: Defines if rows/columns with any (how='any') or all (how='all') missing values should be dropped.




```python
# Drop rows with missing values
print(df.dropna())

# Drop columns with missing values
print(df.dropna(axis=1))
```

### **Filling Missing Values**
- Filling missing values with a constant
- Filling numeric columns with the mean

###**Filling Missing Values**
- Use fillna() to replace missing values with a specific value or method (e.g., forward fill, backward fill).
- Parameters:
 - value: Replace with a specific value.
 - method: Use ffill (forward fill) or bfill (backward fill).

```python
# Fill missing values with a constant
df_filled = df.fillna('Unknown')
print(df_filled)

# Fill with mean, median, or mode (for numeric data)
df['Age'] = df['Age'].fillna(df['Age'].mean())
print(df)
```

### **Replacing Missing Values**
- Replacing NaN with a specific value

```python
print("\nDataFrame after replacing NaN in 'City' column with 'Unknown':")
df['City'] = df['City'].replace(np.nan, 'Unknown')
print(df)
```

### **Hands-on Practice**
Load a sample dataset with missing values and handle them.

```python
data_practice = {
    'Product': ['A', 'B', 'C', 'D'],
    'Price': [100, None, 200, None],
    'Stock': [50, None, 30, 10]
}
df_practice = pd.DataFrame(data_practice)

print("\nPractice Dataset:")
print(df_practice)

# Identify missing data
print("\nIdentifying missing data:")
print(df_practice.isnull())

# Drop missing values
print("\nDropping rows with missing values:")
print(df_practice.dropna())

# Fill missing values with default values
print("\nFilling missing values:")
df_practice['Price'] = df_practice['Price'].fillna(150)
df_practice['Stock'] = df_practice['Stock'].fillna(df_practice['Stock'].mean())
print(df_practice)
```

## Lecture 4: Merging, Joining, and Concatenating DataFrames

### **Merging DataFrames**
- Using `merge()` for relational joins

```python
# Example DataFrames for merging
df1 = pd.DataFrame({
    'ID': [1, 2, 3, 6],
    'Name': [np.nan, 'Bob', 'Charlie', 'Sunny']
})

df2 = pd.DataFrame({
    'ID': [2, 3, 4],
    'Score': [85, 90, 78]
})

# Inner Join
print("\nInner Join:")
print(pd.merge(df1, df2, on='ID'))

# Left Join
print("\nLeft Join:")
print(pd.merge(df1, df2, on='ID'))

# Outer Join
print("\nOuter Join:")
print(pd.merge(df1, df2, on='ID'))
```

### **Joining DataFrames**
- Using `join()` method for index-based joins

```python
df3 = pd.DataFrame({'Score': [85, 90, 78]}, index=['Alice', 'Bob', 'Charlie'])
df4 = pd.DataFrame({'Age': [25, 30, 35]}, index=['Alice', 'Bob', 'David'])

print("\nJoining DataFrames (Inner Join):")
print(df3.join(df4, how='inner'))

print("\nJoining DataFrames (Outer Join):")
print(df3.join(df4, how='outer'))
```

### **Concatenating DataFrames**
- Using `concat()` for stacking DataFrames

```python
df5 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df6 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})

# Vertical Concatenation
print("\nVertical Concatenation:")
print(pd.concat([df5, df6]))

# Horizontal Concatenation
print("\nHorizontal Concatenation:")
print(pd.concat([df5, df6], axis=1))
```

# **4. Hands-on Practice (10 mins)**
**Dataset Practice**


```python
# Create DataFrames
df1 = pd.DataFrame({'ID': [1, 2], 'Name': ['Alice', 'Bob']})
df2 = pd.DataFrame({'ID': [2, 3], 'Score': [85, 90]})

# Merge DataFrames
print(pd.merge(df1, df2, on='ID', how='inner'))

# Concatenate DataFrames
df3 = pd.DataFrame({'Age': [25, 30]}, index=['Alice', 'Bob'])
print(pd.concat([df1.set_index('Name'), df3], axis=1))
```
