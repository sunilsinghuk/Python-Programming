# **Lecture 1: Introduction to NumPy**
##**1. What is NumPy?**
NumPy (Numerical Python) is a powerful Python library used for:

✅ Working with arrays (single and multi-dimensional).

✅ Performing mathematical operations, including:

* Linear algebra (matrix operations, vector calculations).
* Fourier transforms (used in signal processing).
* Statistical and scientific computing.

## **📌 History of NumPy**
* Created in 2005 by Travis Oliphant.
* Open-source and widely used in scientific computing, AI, and machine learning.
* Built on top of C and C++, making it extremely fast compared to standard Python lists.

**✅ Example: Simple NumPy Array Creation**


```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
print(arr)  # Output: [1 2 3 4 5]
print(type(arr))  # Output: <class 'numpy.ndarray'>
```

# **2. Why Use NumPy?**
## **🚀 Speed and Efficiency**
Python has lists that serve the purpose of arrays, but they are slow for numerical operations.

🔹 NumPy arrays (ndarray) are up to 50x faster than Python lists.

🔹 It provides vectorized operations (processing multiple elements at once).

🔹 Uses less memory and optimized CPU operations.


**✅ Example: Speed Comparison (NumPy vs. List)**

```python
import numpy as np
import time

# Python List Computation
py_list = list(range(1, 1000000))
start = time.time()
py_result = [x * 2 for x in py_list]
print("Python List Time:", time.time() - start)

# NumPy Array Computation
np_array = np.array(py_list)
start = time.time()
np_result = np_array * 2  # Vectorized Operation
print("NumPy Array Time:", time.time() - start)
```

➡️ NumPy performs operations significantly faster than lists!

# **3. Why is NumPy Faster Than Lists?**
##**🏎️ Memory Optimization**
* Python lists store references to objects at different memory locations → Slower access
* NumPy arrays store data in a single contiguous block of memory → Faster access

**📌 This is called Locality of Reference in Computer Science.**

**✅ Example: Checking Memory Usage of List vs. NumPy**

➡️ NumPy uses significantly less memory compared to lists!

```python
import numpy as np
import sys

py_list = list(range(1000))
np_array = np.array(py_list)

print("Memory used by Python list:", sys.getsizeof(py_list) * len(py_list), "bytes")
print("Memory used by NumPy array:", np_array.nbytes, "bytes")
```

# **7. Hands-on Practice: NumPy Basics**
✅ **Example 1: Creating NumPy Arrays**

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])  # 1D array
print("NumPy Array:", arr)
```

**✅ Example 2: 2D Array Creation**

```python
import numpy as np

arr2D = np.array([[1, 2, 3], [4, 5, 6]])  # 2D array
print("2D Array:\n", arr2D)
```

**✅ Example 3: Checking Array Properties**

```python
print("Shape:", arr2D.shape)  # (2, 3)
print("Data Type:", arr2D.dtype)  # int
print("Number of Dimensions:", arr2D.ndim)  # 2
```

**✅ Example 4: Performing Operations on Arrays**

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

print("Addition:", arr1 + arr2)  # [5 7 9]
print("Multiplication:", arr1 * arr2)  # [4 10 18]
print("Exponentiation:", arr1 ** 2)  # [1 4 9]
```
